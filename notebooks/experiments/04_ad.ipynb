```python
import numpy as np

def analyze_hands(sequence):
    """
    sequence: (T, 126)
    returns stats about left/right hand usage
    """

    left = sequence[:, :63]
    right = sequence[:, 63:]

    left_active = np.sum(np.any(left != 0, axis=1))
    right_active = np.sum(np.any(right != 0, axis=1))

    total_frames = sequence.shape[0]

    return {
        "total_frames": total_frames,
        "left_active_frames": int(left_active),
        "right_active_frames": int(right_active),
        "left_ratio": left_active / total_frames,
        "right_ratio": right_active / total_frames,
    }
```
